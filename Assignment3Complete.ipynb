## 3) Frequent Itemset Mining with FP‑Growth (min_support = 0.02) (20 pts)
We pivot the data to a **transaction × item** one‑hot table (boolean), then run FP‑Growth.

```python
basket = (bread.groupby('transaction')['item'].apply(list).reset_index().sort_values('transaction')['item'].tolist())

te_b = TransactionEncoder()
hot_b = pd.DataFrame(te_b.fit(basket).transform(basket), columns=te_b.columns_).astype(bool)

freq_b = fpgrowth(hot_b, min_support=0.02, use_colnames=True)
freq_b.sort_values('support', ascending=False)
```
